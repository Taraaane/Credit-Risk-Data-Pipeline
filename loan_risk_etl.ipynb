```python
import pandas as pd
import numpy as np
# Load just a sample to see columns and data types
df_sample = pd.read_csv('accepted_2007_to_2018Q4.csv', nrows=5000 , low_memory=False)

# Inspect the data
print("--- Columns ---")
print(df_sample.columns.tolist())

print("\n--- Basic Info ---")
print(df_sample.info())
```

```python
file_path = 'accepted_2007_to_2018Q4.csv'
selected_cols = [
    'loan_amnt', 'int_rate', 'loan_status', 'fico_range_low', 
    'dti', 'home_ownership', 'annual_inc', 'grade', 'purpose', 'delinq_2yrs'
]

#Chunking
chunk_size = 100000 
chunks = []

for chunk in pd.read_csv(file_path, usecols=selected_cols, chunksize=chunk_size, low_memory=False):
    # Cleaning
    chunks.append(chunk)
df = pd.concat(chunks, axis=0)
print(f"Data loaded successfully. Shape: {df.shape}")
```

```python
# Check basic statistics 
display(df.describe())

#Check value distribution for categorical data 
print("\n--- Distribution of Loan Status ---")
print(df['loan_status'].value_counts())

#Check null values specifically for our risk metrics
print("\n--- Missing Value Report ---")
missing_data = df.isnull().sum()
print(missing_data[missing_data > 0])
```

```python
print("--- Summary Statistics ---")
print(df.describe())

print("\n--- Distribution of Loan Status ---")
print(df['loan_status'].value_counts(normalize=True)) 

print("\n--- Missing Value Report (%) ---")
missing_percentage = (df.isnull().sum() / len(df)) * 100
print(missing_percentage[missing_percentage > 0].sort_values(ascending=False))
```

```python
def map_risk(status):
    if status in ['Fully Paid', 'Current', 'Does not meet the credit policy. Status:Fully Paid']:
        return 1
    else:
        return 0

df_clean = df.dropna().copy()
df_clean.loc[:, 'risk_label'] = df_clean['loan_status'].apply(map_risk)

print("--- Cleaning Complete ---")
```

```python
import sqlite3
import pandas as pd


db_name = 'loan_risk.db'
conn = sqlite3.connect(db_name)

#Accepted Data
print("Processing Accepted Data...")
file_path = 'accepted_2007_to_2018Q4.csv'
selected_cols = ['loan_amnt', 'int_rate', 'loan_status', 'fico_range_low', 'dti', 'home_ownership', 'annual_inc', 'grade', 'purpose', 'delinq_2yrs']

is_first_chunk = True

for chunk in pd.read_csv(file_path, usecols=selected_cols, chunksize=100000, low_memory=False):
    
    chunk_clean = chunk.dropna().copy()
    chunk_clean['risk_label'] = chunk_clean['loan_status'].apply(lambda x: 1 if x in ['Fully Paid', 'Current', 'Does not meet the credit policy. Status:Fully Paid'] else 0)
    chunk_clean['fico_score'] = chunk_clean['fico_range_low']
    chunk_clean['is_accepted'] = 1
    
    
    final_chunk = chunk_clean[['loan_amnt', 'fico_score', 'dti', 'is_accepted']]
    
    
    mode = 'replace' if is_first_chunk else 'append'
    final_chunk.to_sql('combined_loan_data', conn, if_exists=mode, index=False, chunksize=50000)
    is_first_chunk = False
    print("Accepted chunk processed.")

# 2. پردازش و ذخیره Rejected Data
print("Processing Rejected Data...")
reject_cols = ['Amount Requested', 'Risk_Score', 'Debt-To-Income Ratio']

for chunk in pd.read_csv('rejected_2007_to_2018Q4.csv', usecols=reject_cols, chunksize=500000, low_memory=False):
    chunk = chunk.rename(columns={'Amount Requested': 'loan_amnt', 'Risk_Score': 'fico_score', 'Debt-To-Income Ratio': 'dti'})
    chunk['dti'] = chunk['dti'].astype(str).str.replace('%', '', regex=False).astype(float)
    
    chunk_clean = chunk.dropna(subset=['fico_score', 'dti']).copy()
    chunk_clean['is_accepted'] = 0
    
   
    final_chunk = chunk_clean[['loan_amnt', 'fico_score', 'dti', 'is_accepted']]
    
    
    final_chunk.to_sql('combined_loan_data', conn, if_exists='append', index=False, chunksize=50000)
    print("Rejected chunk processed.")

conn.close()
print("DONE")
```
